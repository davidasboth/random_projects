# Find players transferring from Hungarian top division to Premier League directly

Using data from https://www.kaggle.com/datasets/cbhavik/football-transfers-from-199293-to-202122-seasons

```python
import pandas as pd
```

```python
transfers = pd.read_csv("./data/premier-league.csv")
transfers.head()
```

Limit to "in" transfers

```python
transfers_in = transfers[transfers["transfer_movement"] == "in"].copy()
transfers_in.head()
```

Let's see what clubs people come in from

```python
transfers_in[
    transfers_in["club_involved_name"].str.lower().str.strip().str.contains("ferenc")
]
```

OK looks like Hungarian clubs are in but annoyingly they don't have a "league from" field.

Let's manually find all Hungarian clubs because apparently we have time on our hands?!

```python
# transfers_in["club_involved_name"] = (
#    transfers_in["club_involved_name"].str.strip().str.lower()
# )

in_clubs = sorted(transfers_in["club_involved_name"].unique())

pd.DataFrame(in_clubs).to_csv("./data/clubs.csv", index=False)
```

These look like the only ones in the data.

Let's pretty print them!

```python
hungarian_clubs = [
    "debrecen",
    "diósgyőr",
    "eto fc győr",
    "fehérvár",
    "ferencváros",
    "mtk budapest",
    "mtk u19",
    "újpest fc",
    "zalaegerszeg",
]

hungarian_transfers = (
    transfers_in[
        transfers_in["club_involved_name"].str.lower().str.strip().isin(hungarian_clubs)
    ]
    .sort_values(["year", "player_name"])
    .copy()
)

transfer_years = hungarian_transfers["year"].unique()

for year in transfer_years:
    print(f"{year}")
    for row in hungarian_transfers[hungarian_transfers["year"] == year].itertuples():
        print(f"\t{row.player_name} to {row.club_name} from {row.club_involved_name}")
```
